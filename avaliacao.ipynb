```python
#QUESTÃO 01 – Contagem Crescente (2 pontos)
contador = 0
while contador <= 20:
  print(contador)
  contador += 1
```

```python
#QUESTÃO 02 – Soma dos Números (2 pontos)
dig = int(input("digite um numero inteiro positivo: "))

soma = 0
for i in range(1, dig + 1):
    soma += i

print(f"A soma de todos os números de 1 até {dig} é: {soma}")
```

```python
#QUESTÃO 03 – Tabuada (2 pontos)
digito = int (input("digite um número de 1 a 10: "))
for i in range(1, 11):
  print (f"{digito} x {i} = {digito * i}")
```

```python
#QUESTÃO 04 – Fatorial (4 pontos)
numero = int(input("digite um numero inteiro positivo: "))
while numero <= 0:
  numero = int(input("digite um NUMERO INTEIRO POSITIVO"))
fator = 1
for i in range(1, numero+1):
  fator = fator*i
  print(f"fatorial = {fator}")
```

```python
#QUESTÃO 05 – Caixa Eletrônico (5 pontos)
notas = [100, 50, 20, 10, 5, 2, 1]
numero = int(input("Valor do saque: "))
print("")
print("Notas utilizadas: ")
for nota in notas:
  quantidade = numero // nota
  if quantidade > 0:
    print(f"{quantidade} nota(s) de R$ {nota}")
    numero %= nota
```
