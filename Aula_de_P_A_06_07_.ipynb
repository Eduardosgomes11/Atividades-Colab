```
# 3 - programa que simula um cofinho, sendo a meta de R$ 10000

meta = 10000.00
total_acumulado = 0.0
mes = 0
mes_da_meta = None

while total_acumulado < meta:
    mes += 1
    deposito = float(input(f"Informe o valor do depósito do mês {mes}: R$ "))
    total_acumulado += deposito

    if total_acumulado >= meta and mes_da_meta is None:
        mes_da_meta = mes

print("-" * 30)
print(f"Quantos meses foram necessários: {mes}")
print(f"O total acumulado: R$ {total_acumulado:.2f}")
print(f"Em qual mês a meta foi atingida: Mês {mes_da_meta}")
```
