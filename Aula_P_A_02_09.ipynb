```python
#Exercício 1 - criando e acessando uma lista

frutas = ["Maçã", "Banana", "Morango", "Uva", "Abacaxi"]

print("Lista completa:", frutas)
print("Primeiro elemento:", frutas[0])
print("Último elemento:", frutas[4])
print("Segundo e terceiro elementos:", frutas[1], "e", frutas[2])
```

```python
#Exercício 2 - adicionando elementos

alunos = ["Ana", "Carlos", "Maria", "Pedro"]

alunos.append("João")
alunos.insert(0, "Beatriz")
alunos.insert(2, "Lucas")

print("Lista de alunos atualizada:", alunos)
```

```python
#Exercício 3 - removendo elementos

numeros = [10, 20, 30, 40, 50, 60, 70]

numeros.remove(30)
numeros.pop()
numeros.pop(0)

print("Lista final após as remoções:", numeros)
```

```python
#Exercício 4 - soma e média

notas = [7.5, 8.0, 6.5, 9.0, 8.5]

soma_total = sum(notas)
quantidade = len(notas)
media = soma_total / quantidade

maior_nota = max(notas)
menor_nota = min(notas)

print("Soma das notas:", soma_total)
print("Média do aluno:", media)
print("Maior nota:", maior_nota)
print("Menor nota:", menor_nota)

if media >= 7.0:
    print("Resultado: Aluno Aprovado!")
else:
    print("Resultado: Aluno Reprovado!")
```

```python
#Exercício 5 - encontrando números pares

numeros = list(range(1, 21))
pares = []

for num in numeros:
    if num % 2 == 0:
        pares.append(num)

print("Lista apenas com os números pares:", pares)
```

```python
#Exercício 6 - contando elementos

numeros = [2, 5, 2, 8, 2, 10, 5, 3, 2, 7]

print("O número 2 aparece:", numeros.count(2), "vezes")
print("O número 5 aparece:", numeros.count(5), "vezes")

if 10 in numeros:
    print("Sim, o número 10 está na lista!")
else:
    print("Não, o número 10 não está na lista.")
```

```python
#Exercício 7 - separando números negativos e positivos

numeros = [-5, 10, -3, 8, 0, -1, 7, -9, 4]

positivos = []
negativos = []
zeros = []

for num in numeros:
    if num > 0:
        positivos.append(num)
    elif num < 0:
        negativos.append(num)
    else:
        zeros.append(num)

print("Números Positivos:", positivos)
print("Números Negativos:", negativos)
print("Lista de Zeros:", zeros)
```

```python
#Exercício 8 - ordenando uma lista

numeros = [42, 7, 15, 89, 3, 22, 50, 11, 64, 33]
print("Lista do jeito que foi criada:", numeros)

numeros.sort()
print("Em ordem crescente:", numeros)

numeros.sort(reverse=True)
print("Em ordem decrescente:", numeros)
```

```python
#Exercício 9 - cadastro de produtos

produtos = [
    ["Arroz", 25.90],
    ["Feijão", 8.50],
    ["Macarrão", 6.90],
    ["Leite", 5.50],
    ["Café", 15.00]
]

total_compra = 0

print("--- Itens na Prateleira ---")
for item in produtos:
    nome = item[0]
    preco = item[1]
    print("Produto:", nome, "| Preço: R$", preco)
    total_compra = total_compra + preco

print("\nO total da nossa compra deu: R$", total_compra)
```

```python
#Exercício 10 - sistema de notas()

total_alunos = int(input("Quantos alunos você quer cadastrar? "))

nomes = []
notas = []

for i in range(total_alunos):
    nome_digitado = input("Digite o nome do aluno: ")
    nota_digitada = float(input("Digite a nota dele: "))
    nomes.append(nome_digitado)
    notas.append(nota_digitada)

print("\n--- Boletim da Turma ---")
for i in range(total_alunos):
    print("Aluno:", nomes[i], "| Nota:", notas[i])

media_sala = sum(notas) / total_alunos
print("\nA média geral da sala foi:", media_sala)
```
