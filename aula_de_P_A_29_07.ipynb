```python
# programa que verifica se um aluno é menoridade ou maioridade

idade = int (input("digite sua idade: "))

if idade >= 18:
  print("você maior de idade")

else:
  print("você menor de idade")
```

```python
# programa um programa que indentifica se o número é positivo ou negativo

numero = int (input("digite um número inteiro: "))

if numero >= 0:
  print(f"o número {numero} é positivo")

else:
  print(f"o número {numero} é negativo")
```
