```python
#exercicio 2

#crie uma lista vazia chamada frutas
#utilizando o .append() adicione 5 frutas informada pelo usuario
#exiba a lista completa

frutas = []

for i in range(5):
    fruta_informada = input(f"Digite a {i+1}ª fruta: ")
    frutas.append(fruta_informada)

print("Lista completa de frutas:", frutas)
```
