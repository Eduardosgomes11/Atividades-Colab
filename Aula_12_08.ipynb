```python
#exercicio 7

#crie uma lista vazia chamada produtos
#peça para o usuario cadastrar 5 produtos utilizando for
#mostre os produtos numerados

#ex:
#1 - arroz


produtos = []

for i in range(5):
    produto = input(f"Digite o nome do {i+1}º produto: ")
    produtos.append(produto)

print("\n--- Lista de Produtos ---")

for indice, produto in enumerate(produtos, start=1):
    print(f"{indice} - {produto}")
```

```python
#exercicio 9

#crie uma lista vazia chamadfa numero
#peça ao usuario para informar 10 numeros
#depois crie 2 listas vazias chamadas pares e impares
#percorra a lista original e coloque cada numero na lista corresponde

numeros = []

for i in range(10):
    num = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(num)

pares = []
impares = []

for num in numeros:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)

print("\n--- Resultados ---")
print(f"Lista original: {numeros}")
print(f"Lista de Pares: {pares}")
print(f"Lista de Ímpares: {impares}")
```

```python
#exercicio 10

#faça um pequeno sistema para cadastrar as notas de 05 alunos
#o programa devera armazenar as notas em uma lista e depois mostrar:
#todas as notas, a menor, quantidade de alunos com nota > ou = a 6, quantidade de alunos com nota < 6.

notas = []

for i in range(5):
    nota = float(input(f"Digite a nota do {i+1}º aluno (0 a 10): "))
    notas.append(nota)

aprovados = 0
reprovados = 0

for nota in notas:
    if nota >= 6.0:
        aprovados += 1
    else:
        reprovados += 1

menor_nota = min(notas)

print("\n--- Relatório Final ---")
print(f"Todas as notas cadastradas: {notas}")
print(f"A menor nota da turma foi: {menor_nota}")
print(f"Quantidade de alunos com nota >= 6: {aprovados}")
print(f"Quantidade de alunos com nota < 6: {reprovados}")
```
